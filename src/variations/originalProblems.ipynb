```python
def findInd(prob,r):
    i = 0
    for p in r:
        if p['id'] == prob:
            return i
        i += 1
    print('INDEX NOT FOUND FOR PROBLEM ' + prob)
    
def create_json_file(save_dir, id_value, problem, solution, probName):
    orig_save = save_dir / 'original'
    orig_save.mkdir(parents=True, exist_ok=True)
    year = probName[:4]
    id = probName[5:]
    data = {
        "year": year,
        "id": id,
        "problem": problem,
        "solution": solution
    }
    with open(orig_save/f'{probName}.json', 'w') as json_file:
        json.dump(data, json_file, indent=4)

p = Path('/lfs/skampere1/0/ericc27/putnam-math/data/Putnam_MATH_variations_static5/test/variation_1')
save_dir = Path('/lfs/skampere1/0/ericc27/putnam-math/data/Putnam_MATH_variations_static5')
i = 0
probs = []
for j in p.iterdir():
    probs.append(str(j)[90:97])

for prob in probs:
    thisP = r[findInd(prob,r)]
    print(thisP)
    create_json_file(save_dir, thisP['id'], thisP['problem'], thisP['solution'], prob)


```
