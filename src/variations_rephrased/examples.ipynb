```python
import json
from putnam2023 import putnam_2023a1, putnam_2023a2, putnam_2023a4, putnam_2023a6
from putnam2013 import putnam2013_b1
from putnam2006 import putnam_2006a3
from utils import PUTNAM_VARIATIONS

# Define the problems you want variations for
problem_keys = {'2023': ['A1', 'A2', 'A4', 'A6'], '2013': ['B1'], '2006': ['A3']}

# Collect the variation samples
variation_samples = {}
for year, ids in problem_keys.items():
    for problem_id in ids:
        print(f'Generating variations for {year} {problem_id}')
        # Fetch the function corresponding to the problem
        problem_function = PUTNAM_VARIATIONS[year][problem_id]
        # Sample a variation
        variation = problem_function.sample()
        # Store the variation using a unique key
        variation_samples[f'{year} {problem_id}'] = variation
```
